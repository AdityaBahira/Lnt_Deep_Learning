# Task 9: Docker Fundamentals and Container Lifecycle Management
**Course**: Deep Learning Deployment & MLOps  
**Approach**: Using Lightweight Docker Hub Image (`python:3.10-slim`, ~125 MB, < 1GB).

## Section 1: Docker Installation & Architecture Verification

```python
# Verify Docker CLI client & engine version
!docker --version
```

```python
# Display detailed system information and container engine drivers
!docker info
```

## Section 2: Explore Docker Images & Registries (Docker Hub)

```python
# Pull lightweight Python 3.10 slim base image from Docker Hub (~125 MB)
!docker pull python:3.10-slim
```

```python
# List downloaded images in local registry cache
!docker images
```

```python
# Inspect image architecture, environment variables, and layer metadata
!docker inspect python:3.10-slim
```

```python
# Tag image for local runtime deployment
!docker tag python:3.10-slim my_dl_runtime:v1
!docker images
```

## Section 3: Docker Networking & Storage Concepts

```python
# Create persistent volume for model workspace & datasets
!docker volume create dl_storage_vol
!docker volume ls
```

```python
# Create custom isolated bridge network
!docker network create dl_bridge_net
!docker network ls
```

## Section 4: Container Lifecycle Operations

```python
# 1. Run container in detached mode using lightweight image
!docker run -d --name dl_runtime_container -v dl_storage_vol:/workspace --network dl_bridge_net python:3.10-slim python -c "import time, sys; print('Lightweight Deep Learning Runtime Initialized'); sys.stdout.flush(); time.sleep(3600)"
```

```python
# 2. Check active running containers
!docker ps
```

```python
# 3. Execute in-container Python verification
!docker exec dl_runtime_container python -c "import sys, platform; print('Python Version:', sys.version.split()[0]); print('Platform:', platform.platform())"
```

```python
# 4. Read container logs
!docker logs dl_runtime_container
```

```python
# 5. Pause and unpause container processes
!docker pause dl_runtime_container
!docker ps
!docker unpause dl_runtime_container
```

```python
# 6. Monitor live resource statistics
!docker stats dl_runtime_container --no-stream
```

```python
# 7. Stop and Start container
!docker stop dl_runtime_container
!docker ps -a
!docker start dl_runtime_container
```

## Section 5: Resource Cleanup

```python
# Clean up container, volume, network, and local tags
!docker stop dl_runtime_container
!docker rm dl_runtime_container
!docker volume rm dl_storage_vol
!docker network rm dl_bridge_net
!docker rmi my_dl_runtime:v1
```
