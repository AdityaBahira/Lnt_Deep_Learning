### 3. Model Inference Test Runs across Sample Patients

````python
# Define 4 Patient Test Vectors
test_patients = {
    "Patient 1 (Young Healthy)": [25.0, 175.0, 68.0, 22.2, 11000.0, 2100.0, 8.0, 62.0, 112.0, 72.0, 6.0, 0.0, 0.0, 0.0],
    "Patient 2 (Moderate Factors)": [45.0, 168.0, 78.0, 27.6, 6000.0, 2500.0, 6.5, 78.0, 128.0, 82.0, 2.0, 4.0, 0.0, 0.0],
    "Patient 3 (Elevated Risk)": [58.0, 162.0, 85.0, 32.4, 4000.0, 2800.0, 5.5, 88.0, 142.0, 90.0, 1.0, 8.0, 1.0, 0.0],
    "Patient 4 (Critical Conditions)": [65.0, 160.0, 96.0, 37.5, 2000.0, 3200.0, 5.0, 105.0, 162.0, 98.0, 0.0, 14.0, 1.0, 1.0]
}

print("=== Running PyTorch Deep Learning Inferences ===\n")
for name, vector in test_patients.items():
    result = engine.predict([vector])[0]
    print(f"🔹 {name}: Predicted Label = {result['predicted_label']} (Confidence: {result['confidence_score']*100:.1f}%)")
    print(f"   Probabilities: {result['class_probabilities']}\n")
````
